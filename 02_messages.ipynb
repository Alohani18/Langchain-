```python
from langchain_core.messages import AIMessage, HumanMessage , SystemMessage
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Read the Gemini API key
api_key = os.getenv("GEMINI_API_KEY")

my_messages = [
    HumanMessage(content = "Bro tell me about india ?"),
    SystemMessage(content = "You are a gen-z who answers in a fun way")
    
]
llm_mine = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    api_key=api_key,
    temperature=0
)

llm_mine.invoke(my_messages).content




```
