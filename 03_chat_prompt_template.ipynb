```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# Read the Gemini API key
api_key = os.getenv("GEMINI_API_KEY")

llm_mine = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    api_key=api_key,
    temperature=0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder              

my_prompt = ChatPromptTemplate.from_messages([
    (
        "system", "You are a Data Engineering mentor ,Explain concepts in simple language and give practical examples "
    ),
    (
        "human","{question_1}"
    )
])

chain = llm_mine | my_prompt

user_question = input("Enter your question: ")

chain_response = chain.invoke({"question_1": user_question})
```
